```python
import numpy as np
```

## Activation functions
### Sigmoid函數

```python
def sigmoid(x):
    return 1 / (1 + np.exp(-x))
```

### ReLU函數

```python
def relu(x):
    return np.maximum(0, x)
```

### Softmax函數
使用 `axis=-1`，分別計算每一筆資料的三個類別機率。

```python
def softmax(x):
    x = x - np.max(x, axis=-1, keepdims=True)
    return np.exp(x) / np.sum(np.exp(x), axis=-1, keepdims=True)
```

## Loss functions
### Cross entropy error（交叉熵誤差）
將整批交叉熵總和除以資料筆數，得到平均誤差。

```python
def cross_entropy_error(y, t):
    if y.ndim == 1:
        t = t.reshape(1, t.size)
        y = y.reshape(1, y.size)

    batch_size = y.shape[0]
    return -np.sum(t * np.log(y + 1e-7)) / batch_size
```

## 多層神經網路小批次學習
#### 輸入層
`X` 有五筆資料，每筆有六個特徵值；`Y` 每一列為對應的 one-hot 真實標籤，依序屬於第3、1、2、3、1類。

```python
batch_size = 5

X = np.array([
    [ 1.0,  0.5, -1.0,  2.0, 0.0,  1.5],
    [ 0.5, -0.5,  1.0,  0.0, 1.5, -1.0],
    [-1.0,  1.0,  0.5, -0.5, 2.0,  1.0],
    [ 2.0,  0.0,  1.5,  1.0, 0.5, -0.5],
    [ 0.0,  1.5, -0.5,  0.5, 1.0,  2.0],
])

Y = np.array([
    [0, 0, 1],
    [1, 0, 0],
    [0, 1, 0],
    [0, 0, 1],
    [1, 0, 0],
])
```

```python
print("X =")
print(np.array(X))
print("X.shape =", X.shape)
print("Y =")
print(np.array(Y))
print("Y.shape =", Y.shape)
```

#### 第一隱藏層：5個神經元，ReLU
`Z1 = X dot W1 + b1`，`H1 = relu(Z1)`。

```python
W1 = np.array([
    [ 0.2, -0.1,  0.3,  0.0,  0.1],
    [ 0.1,  0.4, -0.2,  0.2,  0.0],
    [-0.3,  0.2,  0.1, -0.1,  0.4],
    [ 0.4, -0.2,  0.0,  0.3, -0.1],
    [ 0.0,  0.1,  0.2, -0.3,  0.2],
    [-0.1,  0.3, -0.4,  0.1,  0.3],
])
b1 = np.array([0.1, -0.2, 0.05, 0.0, -0.1])
```

```python
print("W1 =")
print(np.array(W1))
print("b1 =")
print(np.array(b1))
print("X.shape =", X.shape)
print("W1.shape =", W1.shape)
print("b1.shape =", b1.shape)
```

```python
Z1 = np.dot(X, W1) + b1
print("Z1 =")
print(np.array(Z1))
print("Z1.shape =", Z1.shape)
```

```python
H1 = relu(Z1)
print("H1 =")
print(np.array(H1))
print("H1.shape =", H1.shape)
```

#### 第二隱藏層：4個神經元，ReLU
`Z2 = H1 dot W2 + b2`，`H2 = relu(Z2)`。

```python
W2 = np.array([
    [ 0.3, -0.2,  0.1,  0.4],
    [-0.1,  0.5,  0.2, -0.3],
    [ 0.4,  0.1, -0.2,  0.2],
    [ 0.2, -0.3,  0.4,  0.1],
    [-0.2,  0.2,  0.3, -0.4],
])
b2 = np.array([0.05, -0.1, 0.1, -0.05])
```

```python
print("W2 =")
print(np.array(W2))
print("b2 =")
print(np.array(b2))
print("H1.shape =", H1.shape)
print("W2.shape =", W2.shape)
print("b2.shape =", b2.shape)
```

```python
Z2 = np.dot(H1, W2) + b2
print("Z2 =")
print(np.array(Z2))
print("Z2.shape =", Z2.shape)
```

```python
H2 = relu(Z2)
print("H2 =")
print(np.array(H2))
print("H2.shape =", H2.shape)
```

#### 第三隱藏層：4個神經元，Sigmoid
`Z3 = H2 dot W3 + b3`，`H3 = sigmoid(Z3)`。

```python
W3 = np.array([
    [ 0.2, -0.3,  0.4,  0.1],
    [-0.1,  0.2,  0.3, -0.4],
    [ 0.3,  0.1, -0.2,  0.2],
    [-0.2,  0.4,  0.1,  0.3],
])
b3 = np.array([0.1, -0.1, 0.0, 0.2])
```

```python
print("W3 =")
print(np.array(W3))
print("b3 =")
print(np.array(b3))
print("H2.shape =", H2.shape)
print("W3.shape =", W3.shape)
print("b3.shape =", b3.shape)
```

```python
Z3 = np.dot(H2, W3) + b3
print("Z3 =")
print(np.array(Z3))
print("Z3.shape =", Z3.shape)
```

```python
H3 = sigmoid(Z3)
print("H3 =")
print(np.array(H3))
print("H3.shape =", H3.shape)
```

#### 第四層（輸出層）：3個神經元，Softmax
`Z4 = H3 dot W4 + b4`，`y_hat = softmax(Z4)`。
`y_hat` 每一列為一筆資料對應第1、2、3類的預測機率。

```python
W4 = np.array([
    [ 1.2, -0.4, 0.3],
    [-0.3,  1.3, 0.2],
    [ 0.1, -0.2, 1.0],
    [-0.5,  0.2, 0.7],
])
b4 = np.array([0.1, -0.1, 0.2])
```

```python
print("W4 =")
print(np.array(W4))
print("b4 =")
print(np.array(b4))
print("H3.shape =", H3.shape)
print("W4.shape =", W4.shape)
print("b4.shape =", b4.shape)
```

```python
Z4 = np.dot(H3, W4) + b4
print("Z4 =")
print(np.array(Z4))
print("Z4.shape =", Z4.shape)
```

```python
y_hat = softmax(Z4)
print("y_hat =")
print(np.array(y_hat))
print("y_hat.shape =", y_hat.shape)
print("每筆資料的機率總和 =")
print(np.array(np.sum(y_hat, axis=1)))
```

### Cross entropy error（交叉熵誤差）
比較五筆 `y_hat` 與 `Y`，計算整批資料的平均交叉熵誤差。

```python
errors = -np.sum(Y * np.log(y_hat + 1e-7), axis=1)
print("每筆資料的交叉熵誤差 =")
print(np.array(errors))

loss = cross_entropy_error(y_hat, Y)
print("整批五筆資料的平均交叉熵誤差 =", loss)
```
